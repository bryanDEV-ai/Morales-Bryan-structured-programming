```python
"""
Lab 4. Institutional email generator and Grade Calculator
"""

# Variable declarations
name = input("Ingrese su nombre: ")
lastname = input("Ingrese su apellido: ")

grade1 = float(input("Ingrese la calificación 1: "))
grade2 = float(input("Ingrese la calificación 2: "))
grade3 = float(input("Ingrese la calificación 3: "))

def generate_email(name, lastname):
    email = f"{name.lower()}.{lastname.lower()}@utd.edu.mx"
    return email

def calculate_final_grade(g1, g2, g3):
    final_grade = (g1 + g2 + g3) / 3
    return round(final_grade, 2)

# Main program
generated_email = generate_email(name, lastname)
final = calculate_final_grade(grade1, grade2, grade3)
print("Your institutional email is:", generated_email)
print("Your final grade is:", final)
```
