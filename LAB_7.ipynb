```python
name = input("Name: ")
programming = int(input("Programming: "))
design = int(input("Design: "))
networking = int(input("Networking: "))


student = {
    "name": name,
    "programming": programming,
    "design": design,
    "networking": networking,
}

#Se imprimen los scores dados

print("Name:", student["name"])
print("Programming:", student["programming"])
print("Design:", student["design"])
print("Networking:", student["networking"])

max_score = max(student["programming"], student["design"], student["networking"])

scores = [student["programming"], student["design"], student["networking"]]
highest_count = scores.count(max_score)

if highest_count > 1:
    career = "Multiple Career Paths Identified"
elif student["programming"] == max_score:
    career = "Software Developer"
elif student["design"] == max_score:
    career = "Designer"
elif student["networking"] == max_score:
    career = "Network Administrator"

print("Recommended Career:", career)
```
