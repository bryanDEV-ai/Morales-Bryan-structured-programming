# 1.2 Structured Programming (functions + control structures)

````python
def calculate_metrics(grade_list):
    total_sum = sum(grade_list)
    avg = total_sum / len(grade_list)
    highest = max(grade_list)
    return avg, highest

def determine_status(average_grade):
    if average_grade >= 70:
        return "Approved"
    return "Fail"

def main():
    student_grades = [70, 80, 90, 70, 50, 100, 80, 60]

    avg, highest = calculate_metrics(student_grades)
    final_status = determine_status(avg)
    print(f"Average: {avg}\nHighest: {highest}\nStatus: {final_status}")

main()
````

Reflection: By wrapping the code in functions with simple names, it becomes easier to read. If something goes wrong in the calculation, you can pinpoint which function to check for the error without disrupting the main flow.

````python
%%writefile school_stats.py

def get_average(grades):
    return sum(grades) / len(grades)

def get_top_grade(grades):
    return max(grades)

def check_passing(average, threshold=70):
    return "Approved" if average >= threshold else "Fail"
````

# 4. Object-Oriented Programming (OOP)

````python
class Student:
    def __init__(self, name, grades):
        self.name = name
        self.grades = grades

    def calculate_average(self):
        return sum(self.grades) / len(self.grades)

    def get_highest_grade(self):
        return max(self.grades)

    def get_academic_status(self):
        avg = self.calculate_average()
        return "Approved" if avg >= 70 else "Fail"

student_a = Student("Bryan", [70, 80, 90, 70, 50, 100, 80, 60])

print(f"Average: {student_a.calculate_average()}")
print(f"Highest Grade: {student_a.get_highest_grade()}")
print(f"Final Status: {student_a.get_academic_status()}")
````

# 1.5. Functional Programming (Programación Funcional)

````python
calc_avg = lambda score_list: sum(score_list) / len(score_list)
find_max = lambda score_list: max(score_list)
verify_pass = lambda average: "Approved" if average >= 70 else "Fail"

initial_grades = [70, 80, 90, 70, 50, 100, 80, 60]

final_average = calc_avg(initial_grades)
max_value = find_max(initial_grades)
academic_result = verify_pass(final_average)


print(f"Result: {academic_result} with average of {final_average}. Max score: {max_value}")
````
